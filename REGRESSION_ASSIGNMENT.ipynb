What is Simple Linear Regression?

Simple Linear Regression is a statistical method used to model the relationship between one independent variable (X) and one dependent variable (Y) by fitting a straight line (Y = mX + c).

Key Assumptions of Simple Linear Regression:

Linearity: The relationship between X and Y is linear.

Independence: Observations are independent.

Homoscedasticity: Equal variance of residuals.

Normality: Residuals are normally distributed.


What does the coefficient m represent?

m represents the slope of the line, indicating the change in Y for a one-unit increase in X.

What does the intercept c represent?

c is the value of Y when X=0, providing the baseline level of the dependent variable.

How do we calculate the slope m in Simple Linear Regression?

m=
Var(X)
Cov(X,Y)
 , where Cov is the covariance and Var is the variance.

Purpose of the least squares method:

To minimize the sum of squared residuals (differences between observed and predicted Y) and find the best-fitting line.

Interpretation of R^2:

R^2 measures the proportion of variance in Y explained by X, with values closer to 1 indicating a better fit.

What is Multiple Linear Regression?

A regression method that models the relationship between a dependent variable and two or more independent variables.

Difference between Simple and Multiple Linear Regression:

Simple regression involves one independent variable, while multiple regression uses two or more.

Key Assumptions of Multiple Linear Regression:

Linearity, Independence, Homoscedasticity, and Normality (same as Simple Linear Regression).

No multicollinearity among independent variables.

The model is correctly specified (includes relevant predictors).

What is heteroscedasticity, and its effects?

Heteroscedasticity occurs when residual variance is not constant. It can lead to inefficient estimates and invalid hypothesis tests.

How to address multicollinearity?

Remove highly correlated variables.

Combine correlated variables.

Use regularization techniques (e.g., Ridge or Lasso regression).

Transforming categorical variables:

Common techniques include:

One-hot encoding.

Label encoding.

Creating dummy variables.

Role of interaction terms:

Interaction terms capture the combined effect of two variables on the dependent variable, showing how their interaction influences Y.

Intercept interpretation difference:

In multiple regression, the intercept represents the value of Y when all independent variables are zero.

Significance of the slope:

The slope indicates the strength and direction of the relationship between an independent variable and Y, affecting predictions proportionally.

Context provided by the intercept:

The intercept gives a reference point for the model, describing the expected Y value when predictors are at their baseline levels.

Limitations of R^2:

Does not indicate causation.

Sensitive to adding variables (even irrelevant ones).

Does not measure prediction accuracy directly.

Interpreting a large standard error for a coefficient:
A large standard error suggests high uncertainty in the coefficient estimate, possibly due to multicollinearity or insufficient data.

Identifying heteroscedasticity in residual plots:
Look for patterns or non-random dispersion in residuals. Addressing it is crucial for valid inference and improving model accuracy.

What is the general equation for polynomial regression?

The general equation for polynomial regression is:

Y=c0+c1X+c2X^2+c3X^3
 +⋯+cdX^d

where
d is the degree of the polynomial,
c0,c1,…,cd are the coefficients.


What are the limitations of polynomial regression?

Overfitting: Higher-degree polynomials may fit the training data too well, leading to poor generalization.
Extrapolation Issues: Predictions outside the range of data can be highly unreliable.
Interpretability: Higher-degree models are harder to interpret.
Computational Complexity: As the degree or number of variables increases, computations become more complex.

What methods can be used to evaluate model fit when selecting the degree of a polynomial?

Cross-validation: Splitting the data and evaluating the model on unseen data.

Adjusted R^2:

Takes into account the number of predictors to avoid overfitting.

AIC/BIC:
Information criteria to balance model fit and complexity.
Residual Analysis: Analyzing residual patterns to ensure good fit.
Why is visualization important in polynomial regression?

Helps to identify the relationship between variables and determine the appropriate degree of the polynomial.
Reveals overfitting or underfitting by comparing the fitted curve to the data points.
Aids in interpreting the model and communicating results effectively.

```python
# How is polynomial regression implemented in Python?
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Sample data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([1.2, 1.9, 3.1, 4.8, 6.0])

# Create polynomial features (e.g., degree = 2)
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# Fit the model
model = LinearRegression()
model.fit(X_poly, Y)

# Predictions
Y_pred = model.predict(X_poly)

# Evaluate
mse = mean_squared_error(Y, Y_pred)
print("Mean Squared Error:", mse)
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```
